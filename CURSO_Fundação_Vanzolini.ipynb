```python
inconsistencias = base_gestores[
    (base_gestores["Resultado Final"] != base_gestores["Resultado_Esperado"]) |
    (base_gestores["Emitir Certificado"] != base_gestores["Certificado_Esperado"])
]

inconsistencias.groupby("Diretoria de Ensino").size().sort_values(ascending=False).head(15)
```

```python
total_diretoria = (
    base_gestores
    .groupby("Diretoria de Ensino")
    .size()
    .reset_index(name="Total_Registros")
)

total_diretoria.sort_values("Total_Registros", ascending=False).head(15)
```

```python
# 1. Quantidade de inconsistências por diretoria
qtd_inconsistencias = (
    inconsistencias
    .groupby("Diretoria de Ensino")
    .size()
    .reset_index(name="Inconsistencias")
)

# 2. Junta com o total de registros de cada diretoria
ranking = total_diretoria.merge(
    qtd_inconsistencias,
    on="Diretoria de Ensino",
    how="left"
)

ranking["Inconsistencias"] = ranking["Inconsistencias"].fillna(0)

# 3. Calcula a taxa de inconsistência
ranking["Taxa_Inconsistencia_%"] = (
    ranking["Inconsistencias"] / ranking["Total_Registros"] * 100
)

# 4. Ordena: menor taxa = melhor desempenho
ranking = ranking.sort_values(
    "Taxa_Inconsistencia_%",
    ascending=True
)

ranking.head(10)
```

```python
# Atualiza a diretoria padronizada também na base de inconsistências
inconsistencias["Diretoria_Padronizada"] = (
    inconsistencias["Diretoria de Ensino"]
    .astype(str)
    .str.strip()
    .str.upper()
    .str.replace(r"^D\.E\.REG\.\s*", "", regex=True)
    .str.replace(r"\s+", " ", regex=True)
)

# Total de registros por diretoria
total_diretoria = (
    base_gestores
    .groupby("Diretoria_Padronizada")
    .size()
    .reset_index(name="Total_Registros")
)

# Inconsistências por diretoria
qtd_inconsistencias = (
    inconsistencias
    .groupby("Diretoria_Padronizada")
    .size()
    .reset_index(name="Inconsistencias")
)

# Junta os dados
ranking = total_diretoria.merge(
    qtd_inconsistencias,
    on="Diretoria_Padronizada",
    how="left"
)

ranking["Inconsistencias"] = ranking["Inconsistencias"].fillna(0)

# Taxa de inconsistência
ranking["Taxa_Inconsistencia_%"] = (
    ranking["Inconsistencias"] /
    ranking["Total_Registros"] * 100
)

# Só diretorias com pelo menos 100 registros
ranking = ranking[ranking["Total_Registros"] >= 100]

# Menor taxa = melhor desempenho
ranking = ranking.sort_values("Taxa_Inconsistencia_%")

ranking.head(10)
```

```python
import re

def padronizar_diretoria(x):
    x = str(x).strip().upper()

    # Remove diferentes formatos de prefixos das diretorias
    x = re.sub(
        r'^D\.?\s*E\.?\s*(?:REG(?:IONAL)?|RG)\.?\s*',
        '',
        x
    )

    x = re.sub(
        r'^D\.?\s*ENS\.?\s*(?:REG(?:IONAL)?|RG)\.?\s*',
        '',
        x
    )

    x = re.sub(r'\s+', ' ', x).strip()

    return x

base_gestores["Diretoria_Padronizada"] = (
    base_gestores["Diretoria de Ensino"]
    .apply(padronizar_diretoria)
)

inconsistencias = base_gestores[
    (base_gestores["Resultado Final"] != base_gestores["Resultado_Esperado"]) |
    (base_gestores["Emitir Certificado"] != base_gestores["Certificado_Esperado"])
].copy()

inconsistencias["Diretoria_Padronizada"] = (
    inconsistencias["Diretoria de Ensino"]
    .apply(padronizar_diretoria)
)

total_diretoria = (
    base_gestores
    .groupby("Diretoria_Padronizada")
    .size()
    .reset_index(name="Total_Registros")
)

qtd_inconsistencias = (
    inconsistencias
    .groupby("Diretoria_Padronizada")
    .size()
    .reset_index(name="Inconsistencias")
)

ranking = total_diretoria.merge(
    qtd_inconsistencias,
    on="Diretoria_Padronizada",
    how="left"
)

ranking["Inconsistencias"] = ranking["Inconsistencias"].fillna(0)

ranking["Taxa_Inconsistencia_%"] = (
    ranking["Inconsistencias"] /
    ranking["Total_Registros"] * 100
)

ranking = ranking[ranking["Total_Registros"] >= 100]

ranking = ranking.sort_values(
    ["Taxa_Inconsistencia_%", "Total_Registros"],
    ascending=[True, False]
)

ranking.head(10)
```
